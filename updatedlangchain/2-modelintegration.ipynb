### Streaming

Most Models Can Stream their output content while it is being generated. By displaying output progressively, streaming significantly improves user experience , particularly for longer responses. calling stream() returns an iterator that yeilds output chunks as they are produced. You can use a loop to process each chunk in real-time

```python
model.invoke("Tell me about train no 12860 latest journy")
```

```python
for chunk in model.stream("write the name of the pm till now"):
    print(chunk.text , end="|",flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers"):
    print(chunk.text , end="|",flush=True)
```

### Batch

```python
### Batching A Collection of Indipendnet request to a model can significantlly improve performance and reduce costs,as the processing can be done in parallel
```

```python
responses = model.batch([
    "How are you my ultron",
    "What is dijkastra algorithams",
    "How we can do future travelling with the help of quantumn physics"
])
for response in responses:
    print(response)
```

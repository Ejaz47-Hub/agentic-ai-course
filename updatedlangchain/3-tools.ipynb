### Models

```python
#Models Can request call tools that perform tasks such as fetching data from a database , searching the web, or running code . tools are pairings of:

### 1. A schema, including the name of the tools, a description, and/or aregument definition (often a JSON schema)

### 2. A function or courutine to execute
```

```python
import os
from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-120b")
response = model.invoke("Top 2 Best comdey movies?")
response
```

```python
from langchain.tools import tool
@tool
def get_weather(location:str)-> str:
    """Get The Weather at location"""

model_with_tools=model.bind_tools([get_weather])
```

```python
response = model_with_tools.invoke("What's the weather like in boston?")
print(response)
for tool_call in response.tool_calls:
# view tool calls made by the models
    print(f"Tool:{tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```

### Tool Execution Loops

```python
#step 1 : Model Generates tool call 
messages = [{"role":"user","content":"In which year you have been released"}]
ai_msg=model_with_tools.invoke(messages)

#step 2 : Execute tools and collect results
for tool_call in ai_msg.tool_calls:
    #Execute the tool with the generated arguments
    tool_result=get_weather.invoke(tool_call)
    messages.append(tool_result)

#step 3 : pss results back to model for final response 
final_response = model_with_tools.invoke(messages)
print(final_response.text)
# "The current weather in boston is 72 and sunny"
```
